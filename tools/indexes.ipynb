```python
# Convolution Filter
block_size = 4
shared_width = 6
num_threads = block_size * block_size
total_elts = shared_width * shared_width
total_sum = 0
for local_id in range(num_threads):
    print("Thread: ",local_id)
    sum = 0
    for i in range(local_id, total_elts, num_threads):
        tY = i / shared_width 
        tX = i % shared_width 
        print("i:", i, " ty: ",int(tY)," tx: ",tX, " id:", tY * shared_width + tX) 
        sum += 1
    total_sum += sum
    print("Sum: ",sum, end="\n\n")

print("Total sum: ",total_sum)
```

```python
# Prefix Sum Reduction
stride = 1
bx = 16
while stride < bx:
    print("Stride: ", stride)
    for tid in range(bx):
        index = (tid + 1) * (2 * stride) - 1
        print("Index: ",index)
    stride *= 2
    print()
```

```python
# Prefix Sum Reverse Tree
bx = 16
stride = bx // 4
while stride >= 1:
    print("Stride: ", stride)
    for tid in range(bx):
        index = (tid + 1) * (2 * stride) - 1
        print("Index: ",index, index + stride)
    stride /= 2
    print()
```
