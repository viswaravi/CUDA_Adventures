```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Vector Addition Data
streams = [1, 2, 3, 4]
execution_time = [3.87, 1.94, 1.29, 0.972]  # in milliseconds

# Plot
plt.figure(figsize=(8, 5))
plt.plot(streams, execution_time, marker='o', linestyle='-', color='b')

# Labels and Title
plt.xlabel("Streams")
plt.ylabel("Execution Time (ms)")
plt.title("Execution Time vs Streams")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Show plot
plt.show()
```

```python
# Matmul Data
matrix_size = [128, 256, 512, 1024, 2048]
execution_time_naive_16 = [24.77, 128.58, 940.90, 8.24 * 1000, 61.21 * 1000]  # Block size 16
execution_time_shared_16 = [19.62, 85.63, 607.04, 4.73 * 1000, 39.36 * 1000]
execution_time_naive_32 = [25.09, 124, 707, 5.62 * 1000, 43.97 * 1000]  # Block size 32
execution_time_shared_32 = [21.02, 100.16, 571.23, 4.41 * 1000, 36.14 * 1000]


# Plot
plt.figure(figsize=(8, 5))
plt.plot(matrix_size, execution_time_naive_16, marker='o', linestyle='-', color='r', label='Naive (Block 16)')
plt.plot(matrix_size, execution_time_shared_16, marker='s', linestyle='-', color='b', label='Shared (Block 16)')
plt.plot(matrix_size, execution_time_naive_32, marker='^', linestyle='-', color='g', label='Naive (Block 32)')
plt.plot(matrix_size, execution_time_shared_32, marker='d', linestyle='-', color='m', label='Shared (Block 32)')

# Labels and Title
plt.xlabel("Matrix Size")
plt.ylabel("Execution Time (µs)")
plt.title("Execution Time vs Matrix Size")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Show plot
plt.show()
```
